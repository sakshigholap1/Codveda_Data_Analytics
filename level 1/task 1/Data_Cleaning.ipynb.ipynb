```python
import pandas as pd
from google.colab import files

# Step 1: Create Raw Dataset
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha",
             "Neha", "Rohan", "Amit", "Pooja"],
    "Age": [22, 23, None, 24, 21, None, 22, 25],
    "Salary": [25000, 30000, 35000, None,
               28000, 40000, 25000, None],
    "City": ["PUNE", "mumbai", "Pune", None,
             "MUMBAI", "pune", "PUNE", "Delhi"],
    "Joining_Date": ["01/02/2026", "2026-02-03",
                     "04-02-2026", "05/02/2026",
                     None, "2026-02-06",
                     "01/02/2026", "07/02/2026"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 3: Handle Missing Values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["City"] = df["City"].fillna("Unknown")

# Step 4: Remove Duplicate Rows
df = df.drop_duplicates()

# Step 5: Standardize Data Formats
df["City"] = df["City"].str.strip().str.title()

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df["Joining_Date"] = df["Joining_Date"].fillna(
    df["Joining_Date"].mode()[0]
)

# Step 6: Display Cleaned Dataset
print("\nCleaned Dataset:")
print(df)

# Step 7: Check Missing Values Again
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Step 8: Save Cleaned Dataset
df.to_csv("cleaned_data.csv", index=False)

print("\nCleaned Dataset Saved Successfully!")

# Step 9: Download CSV File
files.download("cleaned_data.csv")
```
