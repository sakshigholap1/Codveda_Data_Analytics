```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset (same as your Task 1)
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha", "Rohan", "Amit", "Pooja"],
    "Age": [22, 23, None, 24, 21, None, 22, 25],
    "Salary": [25000, 30000, 35000, None, 28000, 40000, 25000, None],
    "City": ["PUNE", "mumbai", "Pune", None, "MUMBAI", "pune", "PUNE", "Delhi"],
    "Experience": [1, 2, 3, 1, 4, 5, 1, 3]
}
df = pd.DataFrame(data)

# Cleaning
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["City"] = df["City"].fillna("Unknown")
df["City"] = df["City"].str.strip().str.title()
df = df.drop_duplicates()
df.to_csv('cleaned_data.csv', index=False)

# --- Task 3 Plots ---
plt.style.use('dark_background')

# 1. Bar Plot - Salary by Name
plt.figure(figsize=(8,5))
sns.barplot(x="Name", y="Salary", data=df, palette="viridis")
plt.title("Bar Plot - Salary by Name")
plt.savefig("bar_plot.png")
plt.show()

# 2. Line Chart - Experience
plt.figure(figsize=(8,5))
plt.plot(df["Name"], df["Experience"], marker='o', color='cyan')
plt.title("Line Chart - Experience")
plt.xlabel("Name")
plt.ylabel("Years")
plt.savefig("line_chart.png")
plt.show()

# 3. Scatter Plot - Age vs Salary
plt.figure(figsize=(8,5))
sns.scatterplot(x="Age", y="Salary", data=df, s=100, color='yellow')
plt.title("Scatter Plot - Age vs Salary")
plt.savefig("scatter_plot.png")
plt.show()

print("All 3 plots saved!")
```
