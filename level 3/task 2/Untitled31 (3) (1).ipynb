```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Dataset banavu - Sales Data
np.random.seed(42)
data = {
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 200),
    'Product': np.random.choice(['Laptop', 'Mobile', 'Tablet'], 200),
    'Sales': np.random.randint(1000, 10000, 200),
    'Profit': np.random.randint(100, 2000, 200),
    'Month': np.random.choice(['Jan','Feb','Mar','Apr'], 200)
}
df = pd.DataFrame(data)

# --- HA LINE ADD KELI - CSV BANEL ---
df.to_csv('sales_data.csv', index=False)
print("sales_data.csv banli!")
print(df.head())

# 2. Dashboard - 4 Graphs Ekach Line/Image madhe
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Sales Dashboard - Codveda Task 2', fontsize=16, fontweight='bold')

# Graph 1: Bar Chart - Region wise Sales
df.groupby('Region')['Sales'].sum().plot(kind='bar', ax=axes[0,0], color='skyblue')
axes[0,0].set_title('Sales by Region')

# Graph 2: Line - Month wise Sales
df.groupby('Month')['Sales'].sum().reindex(['Jan','Feb','Mar','Apr']).plot(ax=axes[0,1], marker='o', color='green')
axes[0,1].set_title('Monthly Sales Trend')

# Graph 3: Pie Chart - Product Share
df.groupby('Product')['Sales'].sum().plot(kind='pie', autopct='%1.1f%%', ax=axes[1,0])
axes[1,0].set_title('Product Sales Share')

# Graph 4: Scatter - Sales vs Profit
axes[1,1].scatter(df['Sales'], df['Profit'], alpha=0.6, color='red')
axes[1,1].set_title('Sales vs Profit')
axes[1,1].set_xlabel('Sales')
axes[1,1].set_ylabel('Profit')

plt.tight_layout()
plt.savefig('Dashboard.png') # He image GitHub var tak
plt.show()
```
